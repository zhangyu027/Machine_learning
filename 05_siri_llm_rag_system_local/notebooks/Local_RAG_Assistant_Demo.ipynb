# Personal AI Assistant RAG System

## Project Question

**Can a local RAG assistant answer questions from private documents without sending data to paid cloud APIs?**

This notebook demonstrates:

1. Load local documents  
2. Chunk documents  
3. Build a FAISS vector database  
4. Retrieve relevant chunks  
5. Answer with local Ollama LLM  
6. Show citations/sources  


```
from pathlib import Path
import sys

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(ROOT))

print("Project root:", ROOT)
```

```
from rag.document_loader import load_documents, create_chunks
from rag.vector_store import build_faiss_index, search_index
from rag.ollama_client import ask_ollama
```

```
documents = load_documents(str(ROOT / "data" / "sample_documents"))
chunks = create_chunks(documents)

print("documents:", len(documents))
print("chunks:", len(chunks))
```

```
build_faiss_index(chunks, str(ROOT / "vector_store"))
```

```
question = "Can this assistant answer questions from private documents?"

results = search_index(question, str(ROOT / "vector_store"), top_k=5)

for item in results:
    print(item["rank"], item["filename"], item["chunk_index"], item["score"])
    print(item["text"][:300])
    print("---")
```

```
# Make sure Ollama is running first:
# ollama run llama3.2

answer = ask_ollama(question, results, model_name="llama3.2")
print(answer)
```
